```python
# Activity 6.4
''' Get input from user for a subject.
Calculate the grade using if-elif-else. '''

x = int(input("Enter marks: "))
if x >= 90:
  print("Grade: A")
elif x >= 75:
  print("Grade: B")
elif x >= 60:
  print("Grade: C")
elif x >= 40:
  print("Grade: D")
else:
  print("Grade: F")
```

```python
# Activity 6.5
''' Get input from user for 3 subject and find average.
Calculate the grade using if-elif-else. '''

x = int(input("Enter marks of 1 subject: "))
y = int(input("Enter marks of 2 subject: "))
z = int(input("Enter marks of 3 subject: "))

# Calculate average manually: (sum) / (count)
a = (x + y + z) / 3
print("Average:", a)

if a >= 90:
  print("Grade: A")
elif a >= 75:
  print("Grade: B")
elif a >= 60:
  print("Grade: C")
elif a >= 40:
  print("Grade: D")
else:
  print("Grade: F")
```

```python
# Activity 6.6
''' To check whether the year is a leap year or not.'''

x = int(input("Enter a year: "))
if (x % 400 == 0) or ( x % 4 == 0) and (x % 100 != 0):  # we take these condition fot the exceptional year 1900 ( which is not leap year)
  print("Leap year")
else:
  print("Not a leap year")
```

```python
# Activity 6.9
''' Write a python program to calculate the gross salary of an employee.
If basic<= 10000: HRA = 20%, DA = 80%
If basic<= 20000: HRA = 25%, DA = 90%
Otherwise: HRA + 30%, DA = 95% '''

basic = float(input("Enter basic salary: "))

if basic <= 10000:
  hra = basic * 0.20
  da = basic * 0.80
elif basic <= 20000:
  hra = basic * 0.25
  da = basic* 0.90
else:
   hra = basic * 0.30
   da = basic* 0.95

gross = basic + hra + da
print("Gross Salary:", gross)
```

```python
# Activity 6.10
''' Write a pyhton program to calculate income tax based on annual income:
Up to 2,50,000: No tax
2,50,000 - 5,00,000: 5%
5,00,000 - 10,00,000: 20%
Above 10,00,000: 30% '''

i = float(input("Enter annual income:"))

if i <= 250000:
  tax = 0
elif i <= 500000:
  tax = (i - 250000) * 0.05
elif i <= 1000000:
  tax = (250000 * 0.05) + (i - 500000) * 0.20
else:
  tax = (250000 * 0.05) + (500000 * 0.20) + (i - 1000000) * 0.30

print("Income Tax:", tax)
```
